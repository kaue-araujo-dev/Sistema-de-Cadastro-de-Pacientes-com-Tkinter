```python
import tkinter as tk
# ---------------------------------------------------------
# CRIAÇÃO DA JANELA PRINCIPAL
# ---------------------------------------------------------
# Tk() cria a janela principal da aplicação.
janela = tk.Tk()
# Define o título que aparecerá na parte superior da janela.
janela.title("Cadastro de Pacientes")
# Define o tamanho da janela:
# 400 pixels de largura por 400 pixels de altura.
janela.geometry("400x400")

# ---------------------------------------------------------
# CAMPO: NOME
# ---------------------------------------------------------
# Label é um texto que será mostrado na janela.
tk.Label(janela, text="Nome:").pack()
# Entry cria uma caixa de texto para o usuário digitar.
entrada_nome = tk.Entry(janela)
# pack() posiciona o componente na janela.
entrada_nome.pack()

# ---------------------------------------------------------
# CAMPO: IDADE
# ---------------------------------------------------------
tk.Label(janela, text="Idade:").pack()
entrada_idade = tk.Entry(janela)
entrada_idade.pack()

# ---------------------------------------------------------
# CAMPO: CPF
# ---------------------------------------------------------
tk.Label(janela, text="CPF:").pack()
entrada_cpf = tk.Entry(janela)
entrada_cpf.pack()

# ---------------------------------------------------------
# CAMPO: TIPO SANGUÍNEO
# ---------------------------------------------------------
tk.Label(janela, text="Tipo sanguíneo:").pack()
entrada_sangue = tk.Entry(janela)
entrada_sangue.pack()

# ---------------------------------------------------------
# PARTE DA POO
# ---------------------------------------------------------
# PROGRAMAR_SEM_IA_1:
# Crie a classe Paciente.
#
# A classe deverá possuir:
# - nome
# - idade
# - cpf
# - tipo_sanguineo
#
# Utilize o método construtor __init__().
class Paciente:
    def __init__(self, nome, idade, cpf, tipo_sanguineo):
        self.nome = nome
        self.idade = idade
        self.cpf = cpf
        self.tipo_sanguineo = tipo_sanguineo

# PROGRAMAR_SEM_IA_2:
# Crie uma função chamada cadastrar_paciente().
#
# Essa função deverá:
# 1. Pegar o texto digitado em cada Entry.
# 2. Criar um objeto da classe Paciente.
# 3. Mostrar os dados do objeto na tela.
def cadastrar_paciente():
    nome = entrada_nome.get()
    idade = entrada_idade.get()
    cpf = entrada_cpf.get()
    tipo_sanguineo = entrada_sangue.get()
    
    paciente = Paciente(nome, idade, cpf, tipo_sanguineo)
    
    texto_exibir = f"Nome: {paciente.nome}\nIdade: {paciente.idade}\nCPF: {paciente.cpf}\nTipo: {paciente.tipo_sanguineo}"
    resultado.config(text=texto_exibir)

# PROGRAMAR_SEM_IA_3:
# Crie um botão chamado "Cadastrar paciente".
#
# O botão deverá chamar a função cadastrar_paciente()
# quando o usuário clicar nele.
botao_cadastrar = tk.Button(janela, text="Cadastrar paciente", command=cadastrar_paciente)
botao_cadastrar.pack()

# ---------------------------------------------------------
# ÁREA PARA MOSTRAR O RESULTADO
# ---------------------------------------------------------

# PROGRAMAR_SEM_IA_4:
# Este Label poderá ser utilizado para mostrar os dados
# do paciente depois que o cadastro for realizado.
resultado = tk.Label(janela, text="")
resultado.pack()

# ---------------------------------------------------------
# INICIA A APLICAÇÃO
# ---------------------------------------------------------
# mainloop() mantém a janela aberta e permite que o usuário
# interaja com os componentes da interface.
janela.mainloop()
```
